```python
from nltk.corpus import stopwords
```

```python
stop = stopwords.words('english')
```

```python
stop.extend(['rated', "n", 'nan' , 'x', 'RATED' , 'Rated'])
```

```python
reviews_token_clean = reviews_tokens.apply(lambda each_review : [token for token in each_review if token not in stop] )
```

```python
type(reviews_token_clean)
```

```python
reviews_token_clean
```

```python
total_reviews_2D = list(reviews_token_clean)
```

```python
total_reviews_1D = []

for review in total_reviews_2D:
    for word in review:
       total_reviews_1D.append(word)
```

```python
total_reviews_1D
```

```python
type(total_reviews_1D)
```

```python
from nltk import FreqDist
```

```python
fd = FreqDist()
```

```python
for word in total_reviews_1D:
    fd[word] = fd[word]+1
```

```python
fd.most_common(20)
```

```python
fd.plot(20)
```

```python
from nltk import FreqDist , bigrams, trigrams
```

```python
bi_grams = bigrams(total_reviews_1D)
```

```python
fd_bigrams = FreqDist()

for bigram in bi_grams:
    fd_bigrams[bigram] = fd_bigrams[bigram] + 1
```

```python
fd_bigrams.most_common()
```

```python
fd_bigrams.plot(20)
```

```python
tri_grams = trigrams(total_reviews_1D)
```

```python
fd_trigrams = FreqDist()

for trigram in tri_grams:
    fd_trigrams[trigram] = fd_trigrams[trigram] + 1
```

```python
fd_trigrams.most_common(50)
```

```python
fd_trigrams.plot(30)
```

```python
from geopy.geocoders import Nominatim
```

```python
geolocator = Nominatim(user_agent = "app" , timeout=None)
```

```python
lat = []
lon = []

for location in locations['name']:
    location = geolocator.geocode(location)
    if location is None:
        lat.append(np.nan)
        lon.append(np.nan)
    else:
          lat.append(location.latitude)
          lon.append(location.longitude)


    
```

```python
locations['latitude'] = lat
locations['longitude'] = lon
```

```python
def get_heatmap(cuisine):
    cuisine_df = df[df['cuisines'].str.contains(cuisine)]
    
    cuisine_rescount = cuisine_df['location'].value_counts().reset_index().rename(columns={'location':'name'})
    heatmap_df = cuisine_rescount.merge(locations, on='name', how = 'left')
    print(heatmap_df)

    basemap = folium.Map()
    HeatMap(heatmap_df[['latitude','longitude', "count"]]).add_to(basemap)
    return(basemap)
    
```
